```python
from bikeroot import biobjective_label_correcting_solve
from bikeroot import biobjective_label_setting_solve
from bikeroot import BikeNetwork, CostApplicator, Cost, NgaMobrp_Cost, NGAMOBRPProblem, alg_utils, NGAMOBRPCrossover, NGAMOBRPMutation, NGAMOBRPSampling, NGAMOBRPSurvival, NGAMOBRPSelection, RouteDuplicateElim
from pymoo.termination import get_termination
from pymoo.core.callback import Callback
from pymoo.optimize import minimize

from pymoo.core.evaluator import Evaluator
import osmnx as ox
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.core.population import Population

import numpy as np
bike_net = BikeNetwork("Spokane, WA")
bike_net.load_graph(f'data/spokane_500_cost_vertex_split.graphml')
src = '6357370391'
dst = '12077918977'

bike_net.prints("nodes", 1000)
bike_net.plot()
```
